```python
df['Gejala'] = df['Gejala'].str.replace(r'\s*,\s*', ', ', regex=True)
```

```python
df['Pemeriksaan Laboratorium'] = (
    df['Pemeriksaan Laboratorium']
    .str.replace('positif', 'Positif', regex=False)
)
```

```python
def bersihkan_faktor_risiko(teks):
    bagian = teks.split(',')

    bagian_bersih = []

    for item in bagian:
        item = item.strip().lower()

        if item == 'perokok':
            item = 'Perokok'
        elif item == 'pengguna alkohol':
            item = 'Pengguna Alkohol'
        elif item == 'lingkungan kotor':
            item = 'Lingkungan Kotor'
        elif item == 'lingkungan perkotaan':
            item = 'Lingkungan Perkotaan'
        elif item == 'lingkungan pedesaan':
            item = 'Lingkungan Pedesaan'

        bagian_bersih.append(item)

    return ', '.join(bagian_bersih)

df['Faktor Risiko Lainnya'] = df['Faktor Risiko Lainnya'].apply(bersihkan_faktor_risiko)
```

```python
df['Faktor Risiko Lainnya'].value_counts()
```

```python
for kolom in df.select_dtypes(include='object').columns:
    print("\n===", kolom, "===")
    print(df[kolom].value_counts())
```

```python
df['Durasi Gejala (Hari)'] = (
    df['Durasi Gejala']
    .str.extract(r'(\d+)')[0]
    .astype(int)
)
```

```python
import matplotlib.pyplot as plt

jk = df['Jenis Kelamin'].value_counts()

plt.figure(figsize=(6,4))
jk.plot(kind='bar')

plt.title('Distribusi Jenis Kelamin')
plt.xlabel('Jenis Kelamin')
plt.ylabel('Jumlah Data')
plt.xticks(rotation=0)

for i, v in enumerate(jk):
    plt.text(i, v + 0.3, str(v), ha='center')

plt.show()
```

```python
durasi = df['Durasi Gejala (Hari)'].value_counts().sort_index()

plt.figure(figsize=(6,4))
durasi.plot(kind='bar')

plt.title('Distribusi Durasi Gejala')
plt.xlabel('Durasi Gejala (Hari)')
plt.ylabel('Jumlah Data')
plt.xticks(rotation=0)

for i, v in enumerate(durasi):
    plt.text(i, v + 0.3, str(v), ha='center')

plt.show()
```

```python
riwayat = df['Riwayat Kesehatan'].value_counts()

plt.figure(figsize=(6,4))
riwayat.plot(kind='bar')

plt.title('Distribusi Riwayat Kesehatan')
plt.xlabel('Riwayat Kesehatan')
plt.ylabel('Jumlah Data')
plt.xticks(rotation=0)

for i, v in enumerate(riwayat):
    plt.text(i, v + 0.3, str(v), ha='center')

plt.show()
```
